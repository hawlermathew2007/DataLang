```python
import torch
from torch.utils.data import Dataset
from transformers import BertTokenizerFast

class BertDataset(Dataset):

    def __init__(self, mbert_tokens_list, mbert_labels_list, label2id, max_len=128):
        # Make sure to use BertTokenizerFast so word_ids() works properly
        self.tokenizer = BertTokenizerFast.from_pretrained('bert-base-cased')
        self.tokens_list = mbert_tokens_list
        self.labels_list = mbert_labels_list
        self.label2id = label2id
        self.max_len = max_len

    def __len__(self):
        return len(self.tokens_list)

    def __getitem__(self, idx):
        tokens, labels = self.tokens_list[idx], self.labels_list[idx]

        # Convert string BIO label to integer ID
        label_ids = [self.label2id[label] for label in labels]

        encoding = self.tokenizer(
            tokens,
            is_split_into_words=True,
            padding='max_length',
            truncation=True,
            max_length=self.max_len,
            return_tensors='pt'
        )

        seq_ids = encoding.word_ids(batch_index=0)
        final_labels = []

        previous_word_idx = None
        for seq_id in seq_ids:
            # 1. Special tokens ([CLS], [SEP], [PAD]) get -100
            if seq_id is None:
                final_labels.append(-100)
            
            # 2. If it's the first subword token of a word, assign its true label
            elif seq_id != previous_word_idx:
                if seq_id < len(label_ids):
                    final_labels.append(label_ids[seq_id])
                else:
                    final_labels.append(-100)
                previous_word_idx = seq_id
                
            # 3. If it's a continuing subword piece (e.g., ##oh), set to -100 to ignore it
            else:
                final_labels.append(-100)

        # Structure the dict output for huggingface model
        item = {key: val.squeeze(0) for key, val in encoding.items()}
        item['labels'] = torch.tensor(final_labels, dtype=torch.long)

        return item
```

```python
train_dataset = BertDataset(mbert_tokens_list=train_df['mbert_text_tokens'],
                           mbert_labels_list=train_df['mbert_bio_labels'],
                           label2id=label2id)

train_loader = DataLoader(train_dataset, batch_size=128,
                         shuffle=True)

valid_dataset = BertDataset(mbert_tokens_list=valid_df['mbert_text_tokens'],
                           mbert_labels_list=valid_df['mbert_bio_labels'],
                           label2id=label2id)

valid_loader = DataLoader(valid_dataset, batch_size=128,
                         shuffle=True)
```

```python
from transformers import BertForTokenClassification, get_linear_schedule_with_warmup
from torch.optim import AdamW
```

```python
model = BertForTokenClassification.from_pretrained(
    'bert-base-cased',
    num_labels=len(unique_labels),
    id2label=id2label,
    label2id=label2id
)

save_dir = "pii_mbert"

num_epochs = 10
optimizer = AdamW(model.parameters(),
                 lr=2e-5, weight_decay=0.01)
num_training_steps = len(train_loader) * num_epochs
scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps = int(0.1*num_training_steps),
    num_training_steps=num_training_steps
)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

model.train()
best_loss = float('inf')
epochs_no_improve = 0
n_epochs_stop = 2
```

```python
def train_one_epoch(model, loader):
    model.train()
    total = 0.0
    for batch in loader:
        optimizer.zero_grad()
        out = model(input_ids=batch["input_ids"].to(device),
                    attention_mask=batch["attention_mask"].to(device),
                    labels=batch["labels"].to(device))
        out.loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total += out.loss.item()
    return total / len(loader)

@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total = 0.0
    for batch in loader:
        out = model(input_ids=batch["input_ids"].to(device),
                    attention_mask=batch["attention_mask"].to(device),
                    labels=batch["labels"].to(device))
        total += out.loss.item()
    return total / len(loader)

best_val, no_improve = float("inf"), 0
for epoch in range(num_epochs):
    train_loss = train_one_epoch(model, train_loader)
    val_loss = evaluate(model, valid_loader)
    print(f"Epoch {epoch+1}: train {train_loss:.4f} | val {val_loss:.4f}")

    if val_loss < best_val:
        best_val, no_improve = val_loss, 0
        model.save_pretrained(save_dir)
        train_dataset.tokenizer.save_pretrained(save_dir)
    else:
        no_improve += 1
        if no_improve >= n_epochs_stop:
            print("Early stopping")
            break
            
```
